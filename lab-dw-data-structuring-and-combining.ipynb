2. **Number of Open Complaints (`int` recommended instead of `str`)**

```python
# Customer Lifetime value:

df1['customer_lifetime_value'] = df1['customer_lifetime_value'].map(functions.clean_customer_lifetime_value)
df1['customer_lifetime_value'] = pd.to_numeric(df1['customer_lifetime_value'], errors='coerce')
df2['customer_lifetime_value'] = df2['customer_lifetime_value'].map(functions.clean_customer_lifetime_value)
df2['customer_lifetime_value'] = pd.to_numeric(df2['customer_lifetime_value'], errors='coerce')
df3['customer_lifetime_value'] = df3['customer_lifetime_value'].map(functions.clean_customer_lifetime_value)

print(df1.dtypes)
print(f"\n{df2.dtypes}")
print(f"\n{df3.dtypes}")
```

```python
# Number of open complaints:

# df1 cleaning and transformation
df1['number_of_open_complaints'] = df1['number_of_open_complaints'].map(
    functions.clean_open_complaints
)
df1['number_of_open_complaints'] = df1['number_of_open_complaints'].astype(int)

# df2 -celaning and transformation
df2['number_of_open_complaints'] = df2['number_of_open_complaints'].map(
    functions.clean_open_complaints
)
df2['number_of_open_complaints'] = df2['number_of_open_complaints'].astype(int)

# df3 - cleaning and transformation
df3['number_of_open_complaints'] = df3['number_of_open_complaints'].map(
    functions.clean_open_complaints
)
df3['number_of_open_complaints'] = df3['number_of_open_complaints'].astype(int)
```

```python
# Gender - and insert U, as unkonwn if we have missing data
# I'm not delete these lines because the insured event, the claim, still exists.

# df1 - [nan, 'F', 'M', 'Femal', 'Male', 'female']
df1['gender'] = df1['gender'].replace({
    'Femal': 'F',
    'Male': 'M',
    'female': 'F',
})
df1['gender'] = df1['gender'].fillna('U')

print(df1['gender'].unique())
print(df1['gender'].value_counts(dropna=False))

# df2 - ['F', 'M', nan, 'female', 'Male']
df2['gender'] = df2['gender'].replace({
    'Femal': 'F',
    'Male': 'M',
    'female': 'F',
})
df2['gender'] = df2['gender'].fillna('U')

print(df2['gender'].unique())
print(df2['gender'].value_counts(dropna=False))

# df3 - ['M', 'F'] - not neccessary to cleaning

print(df3['gender'].unique())
print(df3['gender'].value_counts(dropna=False))
```

```python
# State

# df1 - ['Washington',    'Arizona',     'Nevada', 'California',     'Oregon',       'Cali',         'AZ',         'WA',          nan]
# and leave the empty rows
df1['state'] = df1['state'].replace({
    'Cali': 'California',
    'AZ': 'Arizona',
    'WA': 'Washington',
})


print(df1['state'].unique())
print(df1['state'].value_counts(dropna=False))

# df2 - ['Arizona', 'California', 'Oregon', 'Nevada', 'Washington', 'AZ']
df2['state'] = df2['state'].replace({
    'AZ': 'Arizona'
})

print(df2['state'].unique())
print(df2['state'].value_counts(dropna=False))

# df3 - ['Washington', 'Arizona', 'Nevada', 'California', 'Oregon'] - not neccessary to cleaning

print(df3['state'].unique())
print(df3['state'].value_counts(dropna=False))
```

```python
# Education

# df1 - ['Master',             'Bachelor', 'High School or Below', 'College',            'Bachelors',               'Doctor',      nan]
# and leave the empty rows

df1['education'] = df1['education'].replace({
    'Bachelors': 'Bachelor'
})

print(df1['education'].unique())
print(df1['education'].value_counts(dropna=False))

# df2 - ['Bachelor', 'Master', 'College', 'Doctor', 'Bachelors',  'High School or Below']

df1['education'] = df1['education'].replace({
    'Bachelors': 'Bachelor'
})

print(df2['education'].unique())
print(df2['education'].value_counts(dropna=False))

# df3 - ['High School or Below', 'Master', 'Bachelor', 'College', 'Doctor'] - not neccessary to cleaning

print(df3['education'].unique())
print(df3['education'].value_counts(dropna=False))
```

This dataset contains information such as customer demographics, policy details, vehicle information, and the customer's response to the last marketing campaign. Our goal is to explore and analyze this data by performing data cleaning, formatting, and structuring.

```python
import numpy as np
import pandas as pd

url ="https://raw.githubusercontent.com/data-bootcamp-v4/data/main/marketing_customer_analysis_clean.csv"
marketing = pd.read_csv(url)


marketing.shape
display(marketing)

marketing.info()
```

### Column Names Operations

```python
marketing.columns = (
    marketing.columns
    .str.strip()        
    .str.lower()        
    .str.replace(" ", "_")
    .str.replace(":", "")
    
)

marketing = marketing.rename(columns={
    "employmentstatus": "employment_status"
})


marketing = marketing.drop(columns=['unnamed_0'])
# The column named `unnamed:_0` is most likely an index left over from a previous save operation
# as it does not contain any new business-relevant information, I removed it

display(marketing.columns.tolist())
```

### Cleaning invalid values

```python
display(marketing['effective_to_date'].unique())
display(marketing['effective_to_date'].nunique())

# round 2 decimal:
marketing['customer_lifetime_value'] = marketing['customer_lifetime_value'].round(2)
marketing['total_claim_amount'] = marketing['total_claim_amount'].round(2)

```

### Dtype operations

```python
marketing['months_since_last_claim'] = marketing['months_since_last_claim'].astype(int)
marketing['number_of_open_complaints'] = marketing['number_of_open_complaints'].astype(int)
marketing['effective_to_date'] = pd.to_datetime(marketing['effective_to_date'])
# now the format is: 2011-01-30 00:00:00, but I do not want to make more modification, 
# since as I understand if I make more modification (slice, forexample) it can be cause problems
# and this format does not effect the analyzation of data 

print(marketing.dtypes)
```

### Null Values

```python
print("Missing values per column:")
print(marketing.isnull().sum())

# since everywhere is 0, no action needed
```

### Duplicates

```python
duplicates_count = marketing.duplicated().sum()
print(f"Number of duplicated rows before cleaning: {duplicates_count}")
```

```python
# Number of duplicated rows before cleaning: 460

marketing = marketing.drop_duplicates(keep='first').reset_index(drop=True)
duplicates_count = marketing.duplicated().sum()
print(f"Number of duplicated rows after cleaning: {duplicates_count}")
```

```python
# 1. Pivot table for aggregation by sales_channel
revenue_by_channel = marketing.pivot_table(
    index='sales_channel',
    values='total_claim_amount',  # vagy monthly_premium_auto
    aggfunc='sum'
).round(2)

# Sort in descending order by highest revenue
revenue_by_channel = revenue_by_channel.sort_values(by='total_claim_amount', ascending=False)

display(revenue_by_channel)
```

*** Dominance of Agent Channel: The Agent channel generates the highest total claim amount by far (over 1.81 million), indicating that personalized, human-driven sales and customer management handle the largest volume of high-value policies or claims.   

*** Traditional vs. Digital Channels: Traditional offline channels (Agent and Branch) account for the vast majority of the total claim amounts, significantly outperforming remote and digital touchpoints.   

*** Lower Digital Engagement: The Web channel records the lowest total claim amount (~706,600), suggesting lower customer adoption, smaller policy sizes, or simpler products being processed online compared to traditional methods.   

*** Strategic Recommendation: Management should investigate ways to boost digital channel adoption (Web) to capture more online market share, while ensuring continued support for the high-performing Agent and Branch networks.

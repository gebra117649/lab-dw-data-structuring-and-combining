### df1: incorrect data types and fixes:

1. **Customer Lifetime Value (`float` recommended instead of `str`)**
Currently stored as a string (`str`) type because it contains a percentage sign (`%`) or other characters.
Fix: Remove non-numeric characters (e.g., using `.str.replace('%', '')`) and then convert to a floating-point number (`float`).

2. **Number of Open Complaints (`int` recommended instead of `str`)**
Stored as a string (`str`), even though the *number* of complaints should clearly be an integer.
Fix: Handle invalid values ​​or text entries using the `pd.to_numeric(..., errors='coerce')` function, then convert to `int`.

3. Categorical columns (`Customer`, `ST`, `GENDER`, `Education`, `Policy Type`, `Vehicle Class`)
These are currently of type `str` (or `object`), which is appropriate for text/categorical data.
Fix: Standardize the data (e.g., unify case, remove extra whitespace using `.str.strip().str.lower()`). You could also convert these to the `category` data type for better memory efficiency, though storing them as strings (`str`) is also acceptable.

***customer_lifetime_value
-Jelenlegi típus: str (vagy hibásan szövegként beolvasott oszlop)

-Javasolt javítás: Eltávolítani a speciális karaktereket a .str.replace('%', '', regex=False) függvénnyel, majd a típust konvertálni lebegőpontos numerikus típusra a pd.to_numeric(..., errors='coerce') segítségével.
